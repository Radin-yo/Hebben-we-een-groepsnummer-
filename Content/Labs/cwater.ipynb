# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

## Theorie

De toegevoerde elektrische energie is:

$$
E = P \cdot t = U \cdot I \cdot t
$$

Deze energie gaat (idealiter) volledig in het opwarmen van het water:

$$
E = m \cdot c \cdot \Delta T
$$

Door beide vergelijkingen aan elkaar gelijk te stellen, volgt:

$$
c = \frac{U \cdot I \cdot t}{m \cdot \Delta T}
$$

### Betekenis van de symbolen

- $ c $ = soortelijke warmte $ [J/(kg \cdot K)]$
- $ m $ = massa van het water $[kg]$
- $ t $ = tijd $[s]$
- $ \Delta T = T_{\text{eind}} - T_{\text{begin}} $

# Resultaten


````python
# Hier de data en de analyse

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit



# Tijd (in seconden)
t = np.array([0, 180])

metingen = [
    {
        "massa_g": 400.4,
        "spanning": 10.0,
        "stroom": 1.02,
        "T": np.array([20.0, 21.0])
    },
    {
        "massa_g": 401.0,
        "spanning": 14.1,
        "stroom": 1.44,
        "T": np.array([18.3, 20.3])
    },
    {
        "massa_g": 398.1,
        "spanning": 19.9,
        "stroom": 2.03,
        "T": np.array([22.0, 25.6])
    }
]

def lineair_model(t, a, b):
    return a * t + b


tijd = 180  # s
c_waarden = []

plt.figure(figsize=(8, 5))

for i, m in enumerate(metingen, start=1):
    # Curve fit
    popt, _ = curve_fit(lineair_model, t, m["T"])
    a, b = popt

    # Temperatuurverloop (gladde lijn)
    t_fit = np.linspace(0, 180, 100)
    T_fit = lineair_model(t_fit, a, b)

    # Fysica
    massa = m["massa_g"] / 1000  # kg
    vermogen = m["spanning"] * m["stroom"]
    energie = vermogen * tijd
    delta_T = a * tijd

    c = energie / (massa * delta_T)
    c_waarden.append(c)

    # Plot
    plt.scatter(t, m["T"], label=f"Meting {i} data")
    plt.plot(t_fit, T_fit, linestyle="--", label=f"Meting {i} fit")

    print(f"Meting {i}")
    print(f"  dT/dt = {a:.5f} °C/s")
    print(f"  c = {c:.0f} J/(kg·K)\n")

plt.xlabel("Tijd (s)")
plt.ylabel("Temperatuur (°C)")
plt.title("Temperatuurstijging van water")
plt.legend()
plt.grid()
plt.savefig("figures/cwater.png", dpi=450)
plt.show()

c_gemiddeld = np.mean(c_waarden)
c_std = np.std(c_waarden)

print(f"Gemiddelde c = {c_gemiddeld:.0f} J/(kg·K)")
print(f"Standaardafwijking = {c_std:.0f} J/(kg·K)")
````

# Discussie en conclusie

## Discussie

Uit de metingen blijkt dat de temperatuur van het water tijdens het verwarmen ongeveer lineair toeneemt in de tijd. Dit bevestigt de aanname dat het elektrisch vermogen gedurende het experiment vrijwel constant was. Door middel van een lineaire fit van de temperatuur als functie van de tijd kon de temperatuurstijging per seconde worden bepaald, wat een betrouwbaardere methode is dan het gebruik van slechts begin- en eindtemperaturen.

De berekende waarden voor de soortelijke warmte van water liggen in de orde van grootte van de literatuurwaarde van ongeveer 4180 J/(kg·K). De afwijkingen tussen de verschillende metingen kunnen worden verklaard door meerdere systematische en toevallige fouten. Zo is aangenomen dat alle toegevoerde elektrische energie volledig wordt gebruikt voor het opwarmen van het water. In de praktijk gaat een deel van de energie verloren aan de omgeving en aan het opwarmen van de maatbeker, de weerstand en de temperatuursensor.

Daarnaast was de massa van het water niet exact gelijk in alle metingen, wat een kleine invloed heeft op de berekende waarde van de soortelijke warmte. Ook konden variaties in omgevingstemperatuur en beperkte menging van het water bijdragen aan meetonnauwkeurigheden. Ondanks het gebruik van een roerder kan de temperatuur lokaal iets verschillen, wat de gemeten temperatuur beïnvloedt.

In dit experiment is de soortelijke warmte van water bepaald door een waterbad elektrisch te verwarmen en de temperatuurstijging in de tijd te meten. Met behulp van elektrische grootheden en een lineaire fit van de temperatuur als functie van de tijd kon de soortelijke warmte worden berekend.

De gevonden waarden komen redelijk overeen met de bekende literatuurwaarde van 4180 J/(kg·K), wat aangeeft dat de gebruikte methode geschikt is voor een experimentele bepaling van de soortelijke warmte.
